# Análise YouTube – Leitura e Escrita de Dados com PySpark

Este notebook realiza a leitura de arquivos CSV, inferência de esquema, visualização, gravação em formato Parquet e registro de tabela no Spark Catalog.

```python
from pyspark.sql import SparkSession

# Inicializar a sessão do Spark
spark = SparkSession.builder \
    .appName("YouTube Analysis - Leitura e Escrita Parquet") \
    .getOrCreate()

spark
```

## 1 e 2. Ler o arquivo `videos-stats.csv`

```python
# Leitura inicial SEM inferência de esquema (tudo como string)
df_videos = spark.read.csv("videos-stats.csv", header=True)
df_videos
```

## 3. Visualizar os primeiros 8 registros

```python
df_videos.show(8, truncate=False)
```

## 4. Visualizar o esquema do arquivo (sem inferência)

```python
df_videos.printSchema()
```

## 5. Ler novamente o arquivo **inferindo o esquema** e visualizar o esquema

```python
df_videos = spark.read.csv(
    "videos-stats.csv",
    header=True,
    inferSchema=True
)

df_videos.printSchema()
df_videos.show(5, truncate=False)
```

## 6. Salvar o arquivo como `videos-parquet` no formato Parquet (com cabeçalho/esquema preservado)

```python
df_videos.write.mode("overwrite").parquet("videos-parquet")
print("Arquivo videos-parquet salvo com sucesso!")
```

## 7. Ler e visualizar o arquivo `videos-parquet` (com cabeçalho/esquema)

```python
df_videos_parquet = spark.read.parquet("videos-parquet")

df_videos_parquet.printSchema()
df_videos_parquet.show(5, truncate=False)
```

## 8. Salvar o DataFrame como tabela `tb_videos` no banco de dados default do Spark Catalog

```python
df_videos_parquet.write.mode("overwrite").saveAsTable("tb_videos")
print("Tabela tb_videos criada no Spark Catalog!")
```

## 9. Listar as tabelas do Spark Catalog

```python
print("Tabelas disponíveis no Spark Catalog:")
spark.catalog.listTables()
```

## 10. Utilizar Spark SQL para ler a tabela `tb_videos`

```python
df_sql = spark.sql("SELECT * FROM tb_videos LIMIT 10")
df_sql.show(truncate=False)
```

## 11. Ler o arquivo `comments.csv` inferindo o esquema e visualizar

```python
df_comments = spark.read.csv(
    "comments.csv",
    header=True,
    inferSchema=True
)

df_comments.printSchema()
df_comments.show(5, truncate=False)
```

## 12. Salvar o arquivo como `comments-parquet` no formato Parquet (com cabeçalho/esquema preservado)

```python
df_comments.write.mode("overwrite").parquet("comments-parquet")
print("Arquivo comments-parquet salvo com sucesso!")

# Validação final
df_comments_parquet = spark.read.parquet("comments-parquet")
df_comments_parquet.printSchema()
df_comments_parquet.show(5, truncate=False)
```
